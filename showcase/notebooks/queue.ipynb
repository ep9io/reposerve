# The queue at a coffee cart

One barista, one customer at a time. People arrive at rate $\lambda$ per
minute and each takes $1/\mu$ minutes, so the M/M/1 result is:

$$L = \frac{\lambda}{\mu - \lambda}, \qquad W = \frac{L}{\lambda}$$

Fifteen customers an hour and a two minute serve put one person in the
system, counting the customer at the counter, for four minutes each.

````python
arrivals = 1 / 4
service = 1 / 2
people = arrivals / (service - arrivals)
print(f"{people:.1f} people in the system, {people / arrivals:.1f} minutes each")
````
