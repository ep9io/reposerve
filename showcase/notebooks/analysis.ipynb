## Load the data

The cell below reads a month of download counts and prints the total.

````python
weekly = [12, 18, 15, 26, 31, 29, 41, 48, 44, 58, 66, 71]
total = sum(weekly)
mean = total / len(weekly)
print(f'{len(weekly)} weeks, {total} downloads, mean {mean:.1f}')
````

````python
from IPython.display import Latex
Latex(r'\text{growth} = \frac{71 - 12}{12} \times 100\% = 492\%')
````

````python
import pandas as pd
frame = pd.DataFrame({'week': range(1, 5), 'downloads': weekly[:4]})
frame
````

````python
from matplotlib import pyplot as plt

weeks = range(1, 13)
plt.plot(weeks, weekly, marker='o')
plt.fill_between(weeks, weekly, alpha=0.15)
plt.title('Weekly downloads')
plt.tight_layout()
````
